```python
import sympy as sp
from IPython.display import display, Math


#! pip install sympy ipython

class CalculadoraSEP:
    """
    Uma classe para gerar e exibir equações de Sistemas Elétricos de Potência (SEP)
    usando a biblioteca SymPy.
    """

    def __init__(self):
        """
        Inicializa a calculadora com símbolos padrão x, y, z.
        """
        print("-" * 50)
        print("Calculadora SEP")
        # Cria os símbolos iniciais x, y, z
        x, y, z = sp.symbols('x y z', real=True)
        
        # Armazena os símbolos em um dicionário para fácil acesso e adição
        self.simbolos = {
            'x': x,
            'y': y,
            'z': z,
        }
        
        # Define valores padrão para substituição, conforme solicitado
        self.valores_padrao = {
            z: 0
        }
        
        print(f"Símbolos iniciais criados: {list(self.simbolos.keys())}")
        print(f"Valores padrão definidos: z = {self.valores_padrao[z]}")
        print("-" * 50)

    def adicionar_simbolo(self, nome: str, **kwargs):
        """
        Adiciona um novo símbolo ao dicionário de símbolos da classe.

        Args:
            nome (str): O nome do símbolo a ser criado (ex: 'P', 'Q', 'delta').
            **kwargs: Argumentos opcionais para sp.Symbol (ex: real=True, positive=True).
        """
        if nome in self.simbolos:
            print(f"Aviso: O símbolo '{nome}' já existe.")
            return
        
        novo_simbolo = sp.Symbol(nome, **kwargs)
        self.simbolos[nome] = novo_simbolo
        print(f"Símbolo '{nome}' adicionado com sucesso.")

    def _mostrar_equacao(self, titulo: str, equacao):
        """
        Função auxiliar para exibir uma equação formatada com um título.
        """
        display(Math(f'\\textbf{{{titulo}}}'))
        display(equacao)
        print("-" * 50)

    def demonstrar_uso_xyz(self):
        """
        Demonstra o uso dos símbolos x, y, z com a substituição z=0.
        """
        x, y, z = self.simbolos['x'], self.simbolos['y'], self.simbolos['z']
        
        # Cria uma expressão genérica com x, y, z
        expressao_original = x**2 + sp.sqrt(y) + z
        
        # Substitui z por seu valor padrão (0)
        expressao_substituida = expressao_original.subs(self.valores_padrao)
        
        self._mostrar_equacao("Expressão Genérica com x, y, z", sp.Eq(sp.Symbol('f(x,y,z)'), expressao_original))
        self._mostrar_equacao("Expressão com z=0 (Substituído)", sp.Eq(sp.Symbol('f(x,y,0)'), expressao_substituida))

    # --- 5 Fórmulas Principais de Engenharia Elétrica para SEP ---

    def formula_potencia_complexa(self):
        """
        Fórmula 1: Potência Complexa (S) em função da tensão (V) e corrente (I).
        S = V * I* (onde I* é o conjugado complexo da corrente)
        """
        # Definindo símbolos específicos para esta fórmula
        S, V, I = sp.symbols('S V I', complex=True)
        P, Q = sp.symbols('P Q', real=True)
        
        # Equação S = P + jQ
        eq1 = sp.Eq(S, P + sp.I * Q)
        self._mostrar_equacao("1. Potência Complexa (Forma Retangular)", eq1)
        
        # Equação S = V * I*
        eq2 = sp.Eq(S, V * sp.conjugate(I))
        self._mostrar_equacao("1. Potência Complexa (Forma Fasorial)", eq2)
        
    def formula_fluxo_potencia_ativa(self):
        """
        Fórmula 2: Fluxo de Potência Ativa entre duas barras (i, j) em um sistema.
        Aproximação para sistemas de alta tensão onde R << X.
        """
        # Símbolos
        P_ij, V_i, V_j = sp.symbols('P_ij |V_i| |V_j|', real=True, positive=True)
        X_ij = sp.Symbol('X_ij', real=True, positive=True)
        delta_i, delta_j = sp.symbols('delta_i delta_j', real=True)
        
        # Construindo a equação
        termo_direita = (V_i * V_j / X_ij) * sp.sin(delta_i - delta_j)
        eq = sp.Eq(P_ij, termo_direita)
        self._mostrar_equacao("2. Fluxo de Potência Ativa (Aproximação DC)", eq)

    def formula_corrente_curto_circuito(self):
        """
        Fórmula 3: Corrente de Curto-Circuito Simétrico Trifásico.
        A corrente de falta é a tensão pré-falta dividida pela impedância de Thevenin.
        """
        # Símbolos
        I_cc = sp.Symbol('|I_{cc}|', positive=True)
        V_f = sp.Symbol('|V_{f}|', positive=True, comment='Tensão pré-falta')
        Z_th = sp.Symbol('|Z_{Th}|', positive=True, comment='Impedância de Thevenin')
        
        # Equação
        eq = sp.Eq(I_cc, V_f / Z_th)
        self._mostrar_equacao("3. Corrente de Curto-Circuito Simétrico", eq)

    def formula_perdas_potencia_ativa(self):
        """
        Fórmula 4: Perdas de Potência Ativa (Joule) em uma linha de transmissão.
        """
        # Símbolos
        P_perda = sp.Symbol('P_{perda}', real=True)
        R = sp.Symbol('R', real=True, positive=True, comment='Resistência da linha')
        I = sp.Symbol('|I|', real=True, positive=True, comment='Magnitude da corrente')

        # Equação
        eq = sp.Eq(P_perda, R * I**2)
        self._mostrar_equacao("4. Perdas de Potência Ativa na Linha", eq)
        
    def formula_equacao_de_swing(self):
        """
        Fórmula 5: A Equação de Swing, fundamental para estudos de estabilidade transitória.
        """
        # Símbolos
        H = sp.Symbol('H', positive=True, comment='Constante de inércia')
        omega_s = sp.Symbol('omega_s', positive=True, comment='Velocidade síncrona')
        delta = sp.Function('delta')
        t = sp.Symbol('t', real=True, positive=True)
        P_m, P_e = sp.symbols('P_m P_e', real=True, comment='Potência mecânica e elétrica')

        # Lado esquerdo da equação (derivada segunda)
        lado_esquerdo = (2 * H / omega_s) * sp.diff(delta(t), t, 2)
        
        # Lado direito (potência acelerante)
        lado_direito = P_m - P_e
        
        # Equação completa
        eq = sp.Eq(lado_esquerdo, lado_direito)
        self._mostrar_equacao("5. Equação de Swing (Estabilidade Transitória)", eq)
```

```python
# 1. Instanciar a classe
calculadora = CalculadoraSEP()

# 2. Demonstrar o uso dos símbolos x, y, z e a substituição z=0
calculadora.demonstrar_uso_xyz()

# 3. Adicionar um novo símbolo (exemplo)
calculadora.adicionar_simbolo('R_carga', real=True, positive=True)
print(f"Símbolos atuais na calculadora: {list(calculadora.simbolos.keys())}")
print("-" * 30)


# 4. Chamar os métodos para exibir as 5 principais fórmulas de SEP
print("\nEXIBINDO AS 5 PRINCIPAIS FÓRMULAS DE SEP\n" + "="*40)
calculadora.formula_potencia_complexa()
calculadora.formula_fluxo_potencia_ativa()
calculadora.formula_corrente_curto_circuito()
calculadora.formula_perdas_potencia_ativa()
calculadora.formula_equacao_de_swing()
```
