## Reworking dataset for data visualization
### Transforming a dataset to improve data visualization in Tableau using Python and Pandas library

* Used for this Tableau visualization: https://public.tableau.com/app/profile/contato/viz/Renewablesenergyanalysis/Dashboard1
* Original dataset by **Our World in Data**: https://www.kaggle.com/datasets/pralabhpoudel/world-energy-consumption


Datasets can be used just 'out of the box' for dashboards and general data visualizations but most of the time, they are not in the best way to create them. 

Here, I rework the dataset ['World Energy Consumption' created by Our World in Data](https://www.kaggle.com/datasets/pralabhpoudel/world-energy-consumption) for my dashboard in Tableau. I tried doing it with the original format but, in order to have interesting tooltips and more advanced charts, I decided it would be easier to have it in a different way.

Below I show step by step of this rework using Pandas library.


#### Importing the dataset

```python
import os
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))
```

#### Transforming it to Pandas DataFrame

```python
import pandas as pd

df0 = pd.read_csv('/kaggle/input/world-energy-consumption/World Energy Consumption.csv')

df0
```

#### Filtering dataset with the columns I will use only

```python
df0 = df0[['iso_code', 
           'country', 
           'year',
           'biofuel_electricity', 
           'hydro_electricity',
           'nuclear_electricity',
           'solar_electricity',
           'wind_electricity', 
           'other_renewable_electricity',
           'coal_electricity',
           'gas_electricity',
           'oil_electricity',
           'biofuel_elec_per_capita', 
           'hydro_elec_per_capita',
           'nuclear_elec_per_capita',
           'solar_elec_per_capita',
           'wind_elec_per_capita', 
           'coal_elec_per_capita',
           'gas_elec_per_capita',
           'oil_elec_per_capita',
           'population'
          ]]

df0
```

#### Getting a part of the dataset to use as 'control' for our transformations

```python
df0.loc[(df0['country'] == 'Brazil') & (df0['year'] == 2018)]
```

#### Merging different fuels' columns into one column called 'fuel'

```python
df = df0.melt(id_vars=['iso_code', 'country', 'year', 'population'],
              var_name='fuel',
              value_name='production'
)

df
```

#### Checking the result with our 'control'

```python
df.loc[(df['country'] == 'Brazil') & (df['year'] == 2018)]
```

#### Creating the column 'per_capita' to move the values of fuel_per_capita

```python
#df_pc = df.loc[df['type'].str.contains(r'[a-z]+_per_capita')]
#df_elec = df.loc[df['type'].str.contains(r'[a-z]+_electricity')]

def split_type(row):
    if row['fuel'].find('per_capita') != -1:
        value = row['production']
    else:
        value = 0
    return value

df['per_capita'] = df.apply(split_type, axis=1)

df
```

#### Checking it again with our 'control'

```python
df.loc[(df['country'] == 'Brazil') & (df['year'] == 2018)]
```

#### Now cleaning the values in column production that were moved to column per_capita

```python
def clean_production(row):
    if row['per_capita'] != 0:
        value = 0
    else:
        value = row['production']
    return value


df['production'] = df.apply(clean_production, axis=1)

df
```

#### Again, another check...

```python
df.loc[(df['country'] == 'Brazil') & (df['year'] == 2018)]
```

#### Cleaning the names in 'fuel'

```python
df['fuel'] = df['fuel'].str.replace('_elec_per_capita', '')
df['fuel'] = df['fuel'].str.replace('_electricity', '')

df['fuel'].unique()
```

This is just an extra step I realized I need to do to make sure my 'Country' filter would work as intended in the dashboard.

```python
df = df.drop(df[df['country'] == 'World'].index)

df['country'].unique()
```

#### Finally, grouping the dataframe and checking the result with our 'control'

```python
df_final = df.groupby(['iso_code', 'country', 'year', 'population', 'fuel']).sum().reset_index()

df_final.loc[(df_final['country'] == 'Brazil') & (df_final['year'] == 2018)]
```

#### Saving it to csv

```python
df_final.to_csv('reworked_dataset.csv',index=False)
```
