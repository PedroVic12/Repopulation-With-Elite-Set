## Electrical Field of a Charge

An electric field is a region around a charged particle where other charged particles

experience a force. It is a vector field, meaning it has both magnitude and direction.

The electric field at a point in space is defined as the electric force acting on a

positive test charge placed at that point divided by the magnitude of the test charge.

The electric field E  created by a charge can be described mathematically as:

E  = F

q  

\(8.1\)

where F  is the force experienced by a small positive test charge q.

The direction of the electric field is defined as the direction of the force that a

positive test charge would experience. The electric field vector is in the direction

of the force acting on a positive test charge placed in the field. Assuming that the

electric force acts on a positive test charge, the electric field will be in the direction of

the electric force. It points away from positive charges and toward negative charges.



```julia
using Plots

# Constantes

k = 8.99e9  # Constante de Coulomb em N·m²/C²
charge_position = (0,0)

println("Entre com a carga (em Coulombs): ")
#q = parse(Float64, readline())

q = -5e-10  # Carga em Coulombs (negativa para carga negativa, positiva para carga positiva)
println("Carga [C]: $q")

# Grid para o campo elétrico
x = -10:0.5:10
y = -10:0.5:10

# Labels e construção do gráfico e Plot
xlims!(-10, 10)
ylims!(-10, 10)

X = collect(x)
Y = collect(y)

# Calculo dos vetores do campo elétrico
x_grid = X .+ zeros(length(y), length(x))
y_grid = zeros(length(y), length(x)) .+ Y

# quiver plot is used to visualise vector fields, showing direction and magnitude of vectors at specified points in a 2D or 3D space. You can create quiver plots using various plotting libraries, such as Plots.jl or Makie.jl. 

R = sqrt.(x_grid.^2 .+ y_grid.^2)  # Distância do ponto de carga
E_magnitude = k * q ./ R.^2  # Magnitude do campo elétrico

E_magnitude[R .== 0] .= 0  # Evitar divisão por zero no ponto de carga

# Calculo dos componentes do campo elétrico
E_x = E_magnitude .* (x_grid ./ R)
E_y = E_magnitude .* (y_grid ./ R)

println("Campo Elétrico [N/C]: ")
println("E_x: $E_x")
println("E_y: $E_y")




"""
Programme 8.4 The representation of electric field for negative charge
A scatter plot is used to visualise the relationship between two or more variables
by displaying individual data points on a Cartesian coordinate system. You can create
scatter plots using popular plotting libraries such as Plots.jl or Makie.jl.
In Programme 8.4, symbolises vector fields, showing direction and magnitude of
vectors at specified points in a 2D for a negative charge [3].
Programmes 8.5 and 8.6 symbolise the data for a positive charge and electric field
diagram for the positive charge [3].
Programme 8.4 symbolises the electric field vector for a negative charge.
Programme 8.5 shows the output values for positive charge.
Programme 8.6 symbolises the electric field vector for a positive charge.

"""
```

```julia

# Criar o quiver plot
quiver(x_grid, y_grid, quiver=(E_x, E_y), aspect_ratio=1, title="Campo Elétrico de uma Carga Pontual", color=:blue, xlabel="x", ylabel="y")

```

A charge moving with velocity v  in both an electric field E  and a magnetic field B
moves under the influence of the electric force q · E  and the magnetic force q ·v ×B.
The total force acting on the charge is the Lorentz force.


F  = q · E  + q · (v  × B)

(8.4)
Let us consider a uniform electric field directed vertically downwards and a
uniform magnetic field perpendicular to the electric field in the plane of the page.
Assuming that the charge q is positive, a downward electric force and an upward
magnetic force will act on this charge during its motion. If the magnitudes of the
two fields are chosen to be q · E = q · v · B such that the particle travels along a linear horizontal line in the region where the fields are located. From the expression

q · E = q · v · B

is obtained,

v = E/B 

(8.5)

Only particles with this speed can pass through electric and magnetic fields
perpendicular to each other without deviation.
The Lorentz force is the basis of mass spectrometry. The mass spectrometer is a
device that separates masses according to the ratio of their charges. Charged particles
are first passed through a speed selector and then enter a magnetic field. Thanks to
the magnetic force acting on the particle in the magnetic field, the particle draws a
semi-circular trajectory and hits the photographic film [1, 2] (Programmes 8.7 and
8.8).
