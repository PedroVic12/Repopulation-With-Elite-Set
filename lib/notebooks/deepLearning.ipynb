```python
import cv2
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from sklearn.model_selection import train_test_split

def preparar_conjunto_de_dados(caminho_dataset):
    df = pd.read_excel(caminho_dataset)

    # Normalizar o texto
    df['PRODUTOS'] = df['PRODUTOS'].str.lower()
    df['PRODUTOS'] = df['PRODUTOS'].str.replace(' ', '')
    df['PRODUTOS'] = df['PRODUTOS'].str.replace('.', '')
    df['PRODUTOS'] = df['PRODUTOS'].str.replace(',', '')

    # Retornar o conjunto de dados
    return df

# Colete o conjunto de dados de imagens
caminho_dataset =  r'C:\Users\PedroVictorRodrigues\Documents\GitHub\elon-musk\Tecnologia e Inovação\Visão Computacional\assets\Nota fiscal\Nota_fiscal_OCR_template.xlsx'
df_dataset = preparar_conjunto_de_dados(caminho_dataset)


# Prepare o conjunto de dados para o treinamento do modelo
X_train, X_test, y_train, y_test = train_test_split(df_dataset['PRODUTOS'], df_dataset['CATEGORIA'], test_size=0.2, random_state=42)
X_train = X_train.astype('float32') / 255
X_test = X_test.astype('float32') / 255


# Crie o modelo
model = Sequential()
model.add(Conv2D(32, (3, 3), padding='same', activation='relu', input_shape=(28, 28, 3)))
model.add(MaxPooling2D((2, 2)))
model.add(Conv2D(64, (3, 3), padding='same', activation='relu'))
model.add(MaxPooling2D((2, 2)))
model.add(Flatten())
model.add(Dense(128, activation='relu'))
model.add(Dense(10, activation='softmax'))


# Treine o modelo
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.fit(X_train, y_train, epochs=10)


# Faça previsões
y_pred = model.predict(X_test)


# Converta as previsões em um DataFrame do Pandas
df_pred = pd.DataFrame(y_pred, columns=['CATEGORIA'])


# Exiba o DataFrame
print(df_pred)
```
