```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

import pandas as pd

# Definição das funções de otimização
def rosenbrock(x, y):
    return (1 - x)**2 + 100 * (y - x**2)**2

def rastrigin(x, y):
    return 20 + x**2 + y**2 - 10 * (np.cos(2 * np.pi * x) + np.cos(2 * np.pi * y))

def ackley(x, y):
    a = 20
    b = 0.2
    c = 2 * np.pi
    sum_sq_term = -0.5 * (x**2 + y**2)
    cos_term = np.cos(c * x) + np.cos(c * y)
    return -a * np.exp(b * np.sqrt(sum_sq_term / 2)) - np.exp(cos_term / 2) + a + np.e

# Geração dos dados
data = pd.DataFrame(
    {
        'x': np.linspace(-5, 5, 400),
        'y': np.linspace(-5, 5, 400)
    }

)



# Generate meshgrid data
x = np.linspace(-5.15, 5.15, 100)
y = np.linspace(-5.15, 5.15, 100)
X, Y = np.meshgrid(x, y)

# Calculate function values
print(X.shape,Y.shape)



# Cálculo dos valores das funções
z_rosenbrock = rosenbrock(x, y)
z_rastrigin = rastrigin(x, y)
z_ackley = ackley(x, y)

Z_rosenbrock = rosenbrock(X, Y)
Z_rastrigin = rastrigin(X,Y )
Z_ackley = ackley(X, Y)



# Plotting
fig = plt.figure(figsize=(18, 10))



# Rastrigin 2D
ax2 = fig.add_subplot(232)
ax2.contourf(X, Y, Z_rastrigin, levels=50, cmap='viridis')
ax2.plot(x, z_rastrigin, color='blue', label='Rastrigin Line')  # Line plot for Rastrigin
ax2.set_title('Rastrigin Function 2D')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.legend()

# Rosenbrock 2D
ax1 = fig.add_subplot(231)
ax1.contourf(X, Y, Z_rosenbrock, levels=50, cmap='viridis')
ax1.plot(x, z_rosenbrock, color='red', label='Rosenbrock Line')  # Line plot for Rosenbrock
ax1.set_title('Rosenbrock Function 2D')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.legend()

# Rosenbrock 3D
ax4 = fig.add_subplot(234, projection='3d')
ax4.plot_surface(X, Y, Z_rosenbrock, cmap='viridis', edgecolor='none')
ax4.set_title('Rosenbrock Function 3D')
ax4.set_xlabel('X')
ax4.set_ylabel('Y')
ax4.set_zlabel('Z')

# Rastrigin 3D
ax5 = fig.add_subplot(235, projection='3d')
ax5.plot_surface(X, Y, Z_rastrigin, cmap='viridis', edgecolor='none')
ax5.set_title('Rastrigin Function 3D')
ax5.set_xlabel('X')
ax5.set_ylabel('Y')
ax5.set_zlabel('Z')


plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

class DataVisualization:
    def __init__(self, data):
        self.data = data

    def plot_line(self, ax, title="Line Plot"):
        ax.plot(self.data)
        ax.set_title(title)

    def plot_scatter(self, ax, title="Scatter Plot"):
        ax.scatter(range(len(self.data)), self.data)
        ax.set_title(title)

    def plot_bar(self, ax, title="Bar Plot"):
        ax.bar(range(len(self.data)), self.data)
        ax.set_title(title)

    def generate_plots(self):
        fig, axs = plt.subplots(2, 3, figsize=(12, 3))  # 2 rows, 3 cols


        
        self.plot_line(axs[0, 0], "Line Plot")
        self.plot_scatter(axs[0, 1], "Scatter Plot")
        self.plot_bar(axs[0, 2], "Bar Plot")
        
        # Add more plots as needed
        plt.tight_layout()
        plt.show()

# Example usage
data = [1, 2, 3, 4, 5]  # Example data
visualization = DataVisualization(data)
visualization.generate_plots()
```
