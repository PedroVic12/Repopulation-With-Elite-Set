## Flask + SqlAlchemy - MVC

Projeto de Análise de Usinas do SIN com Flask e Streamlit

 arquitetura MVC para processar, armazenar e visualizar dados do Sistema Interligado Nacional (SIN) a partir de um arquivo CSV.

    Backend (Model & Controller): Uma aplicação Flask com SQLAlchemy normaliza os dados do CSV e os armazena em um banco de dados SQLite.

    Frontend (View): Um dashboard interativo em Streamlit que se conecta ao banco de dados para permitir a visualização das tabelas e a execução de consultas SQL personalizadas.

Após o processamento dos dados, você pode iniciar o dashboard interativo.

No terminal, execute o comando do Streamlit:

streamlit run dashboard.py

O dashboard será aberto automaticamente no seu navegador. Você poderá:

    Selecionar qual das tabelas normalizadas deseja visualizar no menu lateral.

    Executar consultas SQL personalizadas para explorar os dados de forma mais aprofundada.

```python
from flask_sqlalchemy import SQLAlchemy

db = SQLAlchemy()

class Usina(db.Model):
    """
    Modelo para a tabela de usinas (dados cadastrais).
    Esta é a tabela principal.
    """
    __tablename__ = 'usinas'
    usina_nome = db.Column(db.String, primary_key=True)
    barra_numero = db.Column(db.Integer)
    estado = db.Column(db.String)
    subsistema = db.Column(db.String)
    tipo_de_usina = db.Column(db.String)
    n_maximo_de_unidades = db.Column(db.Integer)

    # Relacionamentos
    parametros = db.relationship('ParametrosTecnicos', backref='usina', lazy=True, cascade="all, delete-orphan")
    despacho = db.relationship('Despacho', backref='usina', lazy=True, cascade="all, delete-orphan")
    localizacao = db.relationship('Localizacao', backref='usina', lazy=True, cascade="all, delete-orphan")
    sincrono = db.relationship('SincronosInercia', backref='usina', lazy=True, cascade="all, delete-orphan")

class ParametrosTecnicos(db.Model):
    """Modelo para os parâmetros técnicos de cada usina."""
    __tablename__ = 'parametros_tecnicos'
    id = db.Column(db.Integer, primary_key=True)
    usina_nome = db.Column(db.String, db.ForeignKey('usinas.usina_nome'), nullable=False)
    potencia_nominal_mw = db.Column(db.Float)
    pmin_unidade_mw = db.Column(db.Float)
    pmax_unidade_mw = db.Column(db.Float)
    qmin_por_unidade_mvar = db.Column(db.Float)
    qmax_por_unidade_mvar = db.Column(db.Float)
    qmin = db.Column(db.Float)
    qmax = db.Column(db.Float)
    cvu_r_mwh = db.Column(db.Float)
    inflexibilidade = db.Column(db.String)

class Despacho(db.Model):
    """Modelo para os dados de despacho de cada usina."""
    __tablename__ = 'despacho'
    id = db.Column(db.Integer, primary_key=True)
    usina_nome = db.Column(db.String, db.ForeignKey('usinas.usina_nome'), nullable=False)
    despacho_atual = db.Column(db.Float)
    despacho_novo = db.Column(db.Float)
    maquinas_p_ativa_despacho_atual_novo = db.Column(db.String)

class Localizacao(db.Model):
    """Modelo para os dados de localização de cada usina."""
    __tablename__ = 'localizacao'
    usina_nome = db.Column(db.String, db.ForeignKey('usinas.usina_nome'), primary_key=True)
    nome_do_rio = db.Column(db.String)
    nome_da_bacia_estado = db.Column(db.String)
    regiao_de_influencia = db.Column(db.String)

class SincronosInercia(db.Model):
    """Modelo para os dados de máquinas síncronas e inércia."""
    __tablename__ = 'sincronos_inercia'
    id = db.Column(db.Integer, primary_key=True)
    usina_nome = db.Column(db.String, db.ForeignKey('usinas.usina_nome'), nullable=False)
    numero_maximo_de_sincronos_despachaveis = db.Column(db.Float)
    sincronos_despachados = db.Column(db.Float)
    const_inercia_equivalente = db.Column(db.Float)
    rpo_sincronizada = db.Column(db.String)
    pinercia = db.Column(db.Float)



############################################
import pandas as pd
import re

#from models import db, Usina, ParametrosTecnicos, Despacho, Localizacao, SincronosInercia

class DataController:
    """
    Controlador responsável pela lógica de processamento e
    carga dos dados do CSV para o banco de dados.
    """
    def _clean_col_names(self, df):
        """Limpa e padroniza os nomes das colunas de um DataFrame."""
        new_columns = []
        for col in df.columns:
            new_col = str(col).lower().strip()
            new_col = new_col.replace('\n', ' ')
            new_col = re.sub(r'[\s\.]+', '_', new_col)
            new_col = re.sub(r'[^a-z0-9_]+', '', new_col)
            new_col = new_col.strip('_')
            new_columns.append(new_col)
        df.columns = new_columns
        return df

    def process_and_load_data(self, filepath):
        """
        Orquestra o processo completo: ler, limpar, normalizar e carregar no banco.
        """
        print("Iniciando processamento do dataset...")
        try:
            df = pd.read_csv(filepath)
            df = self._clean_col_names(df)

            rename_map = {
                'nome_da_usina': 'usina_nome',
                'numero_da_barra': 'barra_numero'
            }
            df.rename(columns=rename_map, inplace=True)

            df.dropna(how='all', axis=0, inplace=True)
            df.dropna(how='all', axis=1, inplace=True)
            df.dropna(subset=['usina_nome'], inplace=True)

            print("Limpando banco de dados existente...")
            db.drop_all()
            db.create_all()

            processed_usinas = set()

            for index, row in df.iterrows():
                usina_nome = row.get('usina_nome')
                if not usina_nome or usina_nome in processed_usinas:
                    continue

                # Cria a entrada principal da Usina (se for nova)
                usina = Usina(
                    usina_nome=usina_nome,
                    barra_numero=row.get('barra_numero'),
                    estado=row.get('estado'),
                    subsistema=row.get('subsistema'),
                    tipo_de_usina=row.get('tipo_de_usina'),
                    n_maximo_de_unidades=row.get('n_maximo_de_unidades')
                )
                db.session.add(usina)
                processed_usinas.add(usina_nome)

                # Cria entradas nas tabelas relacionadas
                param = ParametrosTecnicos(usina_nome=usina_nome, **row.filter(items=ParametrosTecnicos.__table__.columns.keys()).to_dict())
                desp = Despacho(usina_nome=usina_nome, **row.filter(items=Despacho.__table__.columns.keys()).to_dict())
                local = Localizacao(usina_nome=usina_nome, **row.filter(items=Localizacao.__table__.columns.keys()).to_dict())
                sinc = SincronosInercia(usina_nome=usina_nome, **row.filter(items=SincronosInercia.__table__.columns.keys()).to_dict())

                db.session.add_all([param, desp, local, sinc])

            print("Commitando dados no banco...")
            db.session.commit()
            print("Processamento concluído com sucesso!")

        except Exception as e:
            db.session.rollback()
            print(f"Ocorreu um erro: {e}")
 #           traceback.print_exc()


import os
from flask import Flask
#from models import db
#from controller import DataController

# --- Configuração da Aplicação Flask ---
app = Flask(__name__)

# Configura o caminho do banco de dados na raiz do projeto
#project_folder = os.path.dirname(os.path.abspath(__file__))
database_path = os.path.join("./", "database.db")
app.config['SQLALCHEMY_DATABASE_URI'] = f'sqlite:///{database_path}'
app.config['SQLALCHEMY_TRACK_MODIFICATIONS'] = False

db.init_app(app)

# --- Comandos de CLI para gerenciar a aplicação ---
@app.cli.command("process-data")
def process_data_command():
    """
    Comando para processar o CSV e carregar no banco de dados.
    Uso: flask process-data
    """
    controller = DataController()
    input_file = 'dataset_sin_ons_br.csv' # Arquivo de entrada
    
    if not os.path.exists(input_file):
        print(f"Erro: O arquivo de dados '{input_file}' não foi encontrado.")
        print("Por favor, coloque o arquivo na mesma pasta do 'app.py'.")
        return
        
    with app.app_context():
        controller.process_and_load_data(input_file)

@app.cli.command("create-db")
def create_db_command():
    """Cria as tabelas do banco de dados."""
    with app.app_context():
        db.create_all()
    print("Banco de dados criado.")

# --- Rota Principal (Opcional) ---
@app.route('/')
def index():
    return ("Servidor Flask está rodando. Use os comandos 'flask process-data' "
            "para carregar os dados e 'streamlit run dashboard.py' para ver o dashboard.")

if __name__ == '__main__':
    app.run(debug=True)



import streamlit as st
import pandas as pd
import sqlalchemy
import os

# --- Configuração da Página do Streamlit ---
st.set_page_config(
    page_title="Dashboard de Usinas do SIN",
    page_icon="⚡",
    layout="wide"
)

st.title("⚡ Dashboard de Análise de Usinas do SIN")
st.write("Visualize os dados normalizados do Sistema Interligado Nacional.")

# --- Conexão com o Banco de Dados ---
DB_FILENAME = "database.db"

def get_db_engine():
    """Cria e retorna uma engine de conexão com o banco SQLite."""
    if not os.path.exists(DB_FILENAME):
        st.error(f"Arquivo do banco de dados '{DB_FILENAME}' não encontrado.")
        st.info("Execute o comando 'flask process-data' no terminal para criar e popular o banco de dados.")
        return None
    
    return sqlalchemy.create_engine(f"sqlite:///{DB_FILENAME}")

engine = get_db_engine()

if engine:
    # --- Seleção da Tabela ---
    inspector = sqlalchemy.inspect(engine)
    table_names = inspector.get_table_names()

    st.sidebar.header("Filtros e Opções")
    selected_table = st.sidebar.selectbox(
        "Selecione a tabela para visualizar:",
        options=table_names
    )

    # --- Visualização da Tabela ---
    if selected_table:
        st.header(f"Dados da Tabela: `{selected_table}`")
        try:
            df = pd.read_sql_table(selected_table, engine)
            st.dataframe(df)
            
            st.success(f"{len(df)} registros carregados da tabela '{selected_table}'.")
        except Exception as e:
            st.error(f"Não foi possível carregar a tabela '{selected_table}': {e}")

    # --- Consulta SQL Personalizada ---
    st.header("Consulta SQL Personalizada")
    with st.expander("Clique para expandir e executar uma query SQL"):
        query = st.text_area("Escreva sua consulta SQL aqui:", height=150, value=f"SELECT * FROM {selected_table} LIMIT 10;")
        
        if st.button("Executar Query"):
            if query:
                try:
                    query_df = pd.read_sql_query(query, engine)
                    st.dataframe(query_df)
                    st.success("Query executada com sucesso.")
                except Exception as e:
                    st.error(f"Erro na consulta SQL: {e}")
            else:
                st.warning("Por favor, escreva uma consulta.")
```
